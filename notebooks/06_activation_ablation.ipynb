## Activation Function Ablation

**Fixed:**
- Embedding = BEST_EMBEDDING (from notebook 04)
- VQC = BEST_VQC (from notebook 05)
- Optimizer = Adam (framework default)

**Variable:**
- tanh (current default)
- sigmoid
- relu (added at runtime)
- identity (none)

```python
# ============================================================
# CELL 10 — Extend FeatureCompressor to support ReLU
# ============================================================
# The framework's FeatureCompressor only supports tanh, sigmoid, none.
# We patch it at runtime to also support "relu" for this ablation.

import torch.nn as nn
from models.custom_vqc.feature_compressor import FeatureCompressor

# Add relu to the supported activations
FeatureCompressor._ACTIVATIONS["relu"] = nn.ReLU

print("✅ FeatureCompressor patched: now supports", list(FeatureCompressor._ACTIVATIONS.keys()))
```

```python
# ============================================================
# CELL 11 — Activation Function Comparison Loop
# ============================================================
import numpy as np
import time
from data.loader import load_dataset_for_language
from preprocessing.pipeline import preprocess_dataset
from embeddings.factory import get_embedder, embed_dataset
from models.custom_vqc.hybrid_model import CustomVQCModel, CustomVQCClassifier
from evaluation.metrics import compute_metrics
from visualization.plots import print_results_table, plot_model_comparison

ACTIVATIONS_TO_TEST = ["tanh", "sigmoid", "relu", "none"]
ACTIVATION_LABELS = {"tanh": "tanh", "sigmoid": "sigmoid", "relu": "relu", "none": "identity"}

# Load data ONCE (reuse across activations)
print("Loading and embedding data...")
dataset = load_dataset_for_language(LANGUAGE, cfg, max_samples=MAX_SAMPLES)
dataset = preprocess_dataset(dataset, cfg)
embedder = get_embedder(cfg, LANGUAGE)
embeddings = embed_dataset(embedder, dataset)

X_train = embeddings["train"]
y_train = np.array(dataset.train["labels"])
X_val   = embeddings["val"]
y_val   = np.array(dataset.val["labels"])
X_test  = embeddings["test"]
y_test  = np.array(dataset.test["labels"])

input_dim  = X_train.shape[1]
num_labels = dataset.num_labels

print(f"  Data: train={len(y_train)}, val={len(y_val)}, test={len(y_test)}")
print(f"  Input dim: {input_dim}, Classes: {num_labels}")
print("✅ Data ready.")

# --- Run activation ablation ---
activation_results = {}

for act_name in ACTIVATIONS_TO_TEST:
    label = ACTIVATION_LABELS[act_name]
    print(f"\n{'='*60}")
    print(f"  Testing activation: {label} (internal: {act_name})")
    print(f"{'='*60}")

    try:
        # Build model with specific activation
        import torch
        model = CustomVQCModel(
            input_dim=input_dim,
            n_qubits=N_QUBITS,
            n_layers=N_LAYERS,
            num_classes=num_labels,
            version=BEST_VQC,
        ).to(DEVICE)

        # Override the compressor activation
        from models.custom_vqc.feature_compressor import FeatureCompressor
        model.compressor = FeatureCompressor(input_dim, N_QUBITS, activation=act_name).to(DEVICE)

        # Create classifier wrapper and inject model
        clf = CustomVQCClassifier(cfg, num_labels, input_dim, DEVICE)
        clf.model = model

        # Train
        history = clf.train(X_train, y_train, X_val, y_val)
        test_metrics = clf.evaluate(X_test, y_test)

        activation_results[label] = {
            "history": history,
            "test_metrics": test_metrics,
            "params": clf.get_params_count(),
            "activation": act_name,
        }
        print(f"  ✅ {label}: F1={test_metrics['f1']:.4f}, Acc={test_metrics['accuracy']:.4f}")

    except Exception as e:
        print(f"  ❌ {label} failed: {e}")
        import traceback; traceback.print_exc()
        activation_results[label] = {"error": str(e)}

print("\n" + "="*60)
print("  Activation ablation complete!")
print("="*60)
```

```python
# ============================================================
# CELL 12 — Results Table & Comparison Plots
# ============================================================
import os
from visualization.plots import print_results_table, plot_model_comparison

os.makedirs("results/activation_ablation", exist_ok=True)

print_results_table(activation_results, title="Activation Function Comparison")

plot_model_comparison(
    activation_results, metric="f1",
    title=f"Activation Ablation — F1 ({LANGUAGE})",
    save_path="results/activation_ablation/activation_f1.png"
)
plot_model_comparison(
    activation_results, metric="accuracy",
    title=f"Activation Ablation — Accuracy ({LANGUAGE})",
    save_path="results/activation_ablation/activation_acc.png"
)
```

```python
# ============================================================
# CELL 13 — Training Curves
# ============================================================
from visualization.plots import plot_training_curves

for model_name, data in activation_results.items():
    if isinstance(data, dict) and "history" in data:
        h = data["history"]
        if "train_loss" in h and len(h["train_loss"]) > 1:
            plot_training_curves(
                h,
                title=f"Training Curves — {model_name}",
                save_path=f"results/activation_ablation/curves_{model_name}.png"
            )
```

```python
# ============================================================
# CELL 14 — Determine Best Activation
# ============================================================
import json

best_f1 = -1
BEST_ACTIVATION = "tanh"

for name, data in activation_results.items():
    if isinstance(data, dict) and "test_metrics" in data:
        f1 = data["test_metrics"].get("f1", 0)
        if f1 > best_f1:
            best_f1 = f1
            BEST_ACTIVATION = name

print(f"\n{'='*60}")
print(f"  BEST ACTIVATION: {BEST_ACTIVATION} (F1={best_f1:.4f})")
print(f"{'='*60}")

results_summary = {
    "best_activation": BEST_ACTIVATION,
    "best_f1": best_f1,
    "embedding_used": BEST_EMBEDDING,
    "vqc_used": BEST_VQC,
    "all_results": {
        k: v.get("test_metrics", {}) if isinstance(v, dict) else {}
        for k, v in activation_results.items()
    }
}

with open("results/activation_ablation/summary.json", "w") as f:
    json.dump(results_summary, f, indent=2, default=str)

print(f"\nResults saved to: results/activation_ablation/")
print(f"\nUse BEST_ACTIVATION = \"{BEST_ACTIVATION}\" in subsequent notebooks.")
```
