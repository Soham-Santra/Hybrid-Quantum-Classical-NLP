```python
# ============================================================
# CELL 1 — Installation  (then Runtime → Restart Runtime)
# ============================================================
import subprocess, sys

def run(cmd):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if r.stdout.strip(): print(r.stdout.strip())
    return r

print("Step 1/6: Wiping stale packages from disk...")
run("pip uninstall -y pennylane pennylane-lightning autoray jax jaxlib 2>/dev/null || true")
run("find /usr/local/lib/python3.12/dist-packages -maxdepth 1 -name 'pennylane*' -exec rm -rf {} + 2>/dev/null || true")
run("find /usr/local/lib/python3.12/dist-packages -maxdepth 1 -name 'PennyLane*' -exec rm -rf {} + 2>/dev/null || true")
run("find /usr/local/lib/python3.12/dist-packages -maxdepth 1 -name 'autoray*'   -exec rm -rf {} + 2>/dev/null || true")
print("  ✓ Stale files wiped")

print("Step 2/6: Installing numpy + scipy (ABI-compatible pair)...")
run("pip install -q --no-cache-dir numpy==1.26.4")
run("pip install -q --no-cache-dir scipy==1.13.1")
print("  ✓ numpy/scipy OK")

print("Step 3/6: Pinning JAX to a pennylane-compatible version...")
# JAX 0.4.26 is the last version where jax.core.Primitive still exists
run("pip install -q --no-cache-dir 'jax==0.4.26' 'jaxlib==0.4.26'")
print("  ✓ JAX pinned")

print("Step 4/6: Installing autoray + pennylane (matched pair)...")
run("pip install -q --no-cache-dir autoray==0.6.12")
run("pip install -q --no-cache-dir pennylane==0.38.0 pennylane-lightning==0.38.0")
print("  ✓ pennylane/autoray OK")

print("Step 5/6: Installing NLP/ML stack...")
run("pip install -q --no-cache-dir "
    "numba>=0.57.1 "
    "transformers>=4.40.0 "
    "datasets>=2.19.0 "
    "scikit-learn>=1.4.0 "
    "gensim>=4.3.0 "
    "nltk>=3.8.0 "
    "seaborn>=0.13.0 "
    "tabulate>=0.9.0 "
    "pyyaml>=6.0 "
    "pandas==2.2.2 "
    "requests==2.32.4 "
    "'rich>=12.4.4,<14' "
    "'cachetools>=4.2.1,<7' "
    "'pillow<12.0' "
    "'tomlkit>=0.12.0,<0.14.0'"
)
print("  ✓ NLP/ML stack OK")

print("Step 6/6: Verifying installed versions...")
for pkg in ["pennylane", "autoray", "numpy", "scipy", "jax"]:
    run(f"pip show {pkg} | grep -E 'Name|Version'")

print()
print("=" * 60)
print("✅ Done! NOW: Runtime → Restart Runtime")
print("   Then run Cell 2 onwards.")
print("=" * 60)
```

```python
# ============================================================
# CELL 2 — Verify Imports  (AFTER restarting runtime)
# ============================================================
import importlib

print("Verifying all imports...")
print("-" * 50)

ok = True
checks = [
    ("numpy",       "1.26"),
    ("scipy",       "1.13"),
    ("jax",         "0.4.26"),
    ("autoray",     "0.6"),
    ("pennylane",   "0.38"),
    ("torch",       None),
    ("transformers",None),
    ("sklearn",     None),
]

for mod, prefix in checks:
    try:
        m   = importlib.import_module(mod)
        ver = getattr(m, "__version__", "?")
        chk = prefix is None or ver.startswith(prefix)
        print(f"  {'✅' if chk else '⚠️ '} {mod:<15} {ver}")
        if not chk: ok = False
    except Exception as e:
        print(f"  ❌ {mod:<15} {e}")
        ok = False

# Deep pennylane circuit test
print()
try:
    import pennylane as qml
    dev = qml.device("default.qubit", wires=2)
    @qml.qnode(dev)
    def _t():
        qml.Hadamard(wires=0)
        qml.CNOT(wires=[0,1])
        return qml.expval(qml.PauliZ(0))
    print(f"  ✅ pennylane circuit  result={float(_t()):.4f}")
except Exception as e:
    print(f"  ❌ pennylane circuit  {e}")
    ok = False

print()
if ok:
    print("✅ All checks passed — safe to continue!")
else:
    print("❌ Checks failed.")
    print("   → Runtime → Factory Reset Runtime, then rerun Cell 1.")
```

```python
# ============================================================
# CELL 5 — Configuration
# ============================================================
import torch

LANGUAGE         = "english"    # "english", "hindi", "bengali"
EMBEDDING        = "tfidf"      # "tfidf", "bert", "mbert", "indicbert"
N_QUBITS         = 4             # 4, 6, or 8
N_LAYERS         = 2
DEVICE           = "cuda" if torch.cuda.is_available() else "cpu"
MAX_SAMPLES      = None          # None = full dataset, e.g. 500 for quick test
CLASSICAL_EPOCHS = 20
QUANTUM_EPOCHS   = 30

print(f"Language:    {LANGUAGE}")
print(f"Embedding:   {EMBEDDING}")
print(f"Qubits:      {N_QUBITS}")
print(f"Device:      {DEVICE} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'})")
print(f"Max samples: {MAX_SAMPLES or 'Full dataset'}")
```

```python
# ============================================================
# CELL 11 — Experiment 2b: Template VQC (all 3 circuits)
# ============================================================
from experiments.runner import run_quantum_template_experiment
from visualization.plots import print_results_table, plot_model_comparison

all_template_results = {}
for circuit_type in ["strongly_entangling", "basic_entangler", "efficient_su2"]:
    print(f"Testing: {circuit_type}")
    cfg.quantum_template.circuit_type = circuit_type
    result = run_quantum_template_experiment(cfg, languages=[LANGUAGE], max_samples=MAX_SAMPLES)
    if LANGUAGE in result:
        all_template_results.update(result[LANGUAGE])

print_results_table(all_template_results, title="All Template Circuits")
plot_model_comparison(all_template_results, metric="f1", title="Template VQC — Circuit Comparison")
```

```python
# ============================================================
# CELL 12 — Experiment 3: Custom VQC (v1, v2, v3)
# ============================================================
from experiments.runner import run_custom_vqc_experiment
from visualization.plots import print_results_table, plot_model_comparison

all_vqc_results = {}
for version in ["v1", "v2", "v3"]:
    print(f"Running Custom VQC {version}...")
    cfg.custom_vqc.version = version
    result = run_custom_vqc_experiment(cfg, languages=[LANGUAGE], max_samples=MAX_SAMPLES)
    if LANGUAGE in result:
        for ver, data in result[LANGUAGE].items():
            all_vqc_results[f"CustomVQC_{ver}"] = data

print_results_table(all_vqc_results, title="Custom VQC — All Versions")
plot_model_comparison(all_vqc_results, metric="f1", title="Custom VQC — Version Comparison")
```

```python
# ============================================================
# CELL 13 — Experiment 4: Full Comparison (all models)
# ============================================================
from experiments.runner import run_all_experiments

print(f"Running ALL experiments — {LANGUAGE}. This will take a while!")
all_results = run_all_experiments(cfg, languages=[LANGUAGE], max_samples=MAX_SAMPLES)
print("✅ All experiments complete.")
```

```python
# ============================================================
# CELL 14 — Grand Comparison Table
# ============================================================
import os
from visualization.plots import print_results_table, plot_model_comparison

combined = {}
for exp_type, exp_data in all_results.items():
    if isinstance(exp_data, dict) and "error" not in exp_data and LANGUAGE in exp_data:
        for model, data in exp_data[LANGUAGE].items():
            if isinstance(data, dict) and "test_metrics" in data:
                combined[f"{exp_type}/{model}"] = data

os.makedirs("results/figures", exist_ok=True)
print_results_table(combined, title="Grand Comparison — All Models")
plot_model_comparison(combined, metric="f1",       title=f"All Models — F1 ({LANGUAGE})",       save_path="results/figures/grand_comparison_f1.png")
plot_model_comparison(combined, metric="accuracy", title=f"All Models — Accuracy ({LANGUAGE})", save_path="results/figures/grand_comparison_acc.png")
```

```python
# ============================================================
# CELL 15 — Training Curves
# ============================================================
from visualization.plots import plot_training_curves

for model_name, data in combined.items():
    if isinstance(data, dict) and "history" in data:
        h = data["history"]
        if "train_loss" in h and len(h["train_loss"]) > 1:
            plot_training_curves(h,
                title=f"Training Curves — {model_name}",
                save_path=f"results/figures/curves_{model_name.replace('/', '_')}.png")
```

```python
# ============================================================
# CELL 16 — Qubit Count Comparison (4 / 6 / 8)
# ============================================================
from experiments.runner import run_quantum_template_experiment
from visualization.plots import print_results_table, plot_model_comparison

qubit_results = {}
for n_q in [4, 6, 8]:
    print(f"Testing {n_q} qubits...")
    cfg.quantum_template.n_qubits     = n_q
    cfg.quantum_template.circuit_type = "strongly_entangling"
    result = run_quantum_template_experiment(cfg, languages=[LANGUAGE], max_samples=MAX_SAMPLES)
    if LANGUAGE in result:
        for ct, data in result[LANGUAGE].items():
            qubit_results[f"{n_q}_qubits"] = data

print_results_table(qubit_results, title="Qubit Count Comparison")
plot_model_comparison(qubit_results, metric="f1", title="Qubit Count — F1 Comparison")
```
